```python
from pyspark.sql import SparkSession
spark= SparkSession.builder.appName("Create").getOrCreate()
```

```python
#In PySpark, select() function is used to select single, multiple, column by index, all columns from the list and the nested columns from a DataFrame, PySpark select() is a transformation function hence it returns a new DataFrame with the selected columns.
```

```python
# Using Data Frame
```

```python
df_select= spark.read.table("samples.bakehouse.sales_customers")
#display(df_select)
display(df_select.limit(10))
```

```python
df_final = df_select.select("customerID","first_name","last_name","country")
display(df_final)
```

```python
#SQL
```

```python
df_sql= spark.read.table("samples.bakehouse.sales_customers")
df_sql.createOrReplaceTempView("sales_customers")
df_final= spark.sql("select customerID,first_name,last_name,country from sales_customers")
display(df_final)
```

```python
# Select columns by different ways
df_final.select("first_name","last_name").display()
```

```python
df_final.select(df_final.first_name,df_final.last_name).display()
```

```python
df_final.select(df_final["first_name"],df_final["last_name"]).display()
```

```python
# By using col() function
from pyspark.sql.functions import col
df_final.select(col("first_name"),col("last_name")).display()
```

```python

# Select columns by regular expression
df_final.select(df_final.colRegex("`^.*name*`")).display()
```
