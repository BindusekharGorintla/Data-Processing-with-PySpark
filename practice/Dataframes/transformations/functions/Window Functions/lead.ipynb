```python
# Lead() window function is used to look ahead in your dataset and access values from subsequent rows without needing a self-join. It’s particularly useful for comparing the current row with the next one (or several ahead).
# LEAD(return_value [, offset [, default]]) OVER (PARTITION BY expr ORDER BY expr)
```

```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName("window").getOrCreate()
from pyspark.sql.functions import *
from pyspark.sql.window import Window
```

```python
df= spark.read.table("workspace.practice.order_window")
```

```python
windowspec = Window.partitionBy("department").orderBy(col("order_date").desc())
df1 = df.withColumn("new_column",lead(col("amount")).over(windowspec))
display(df1)
```
