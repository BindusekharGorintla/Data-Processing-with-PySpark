```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Joins").getOrCreate()
```

```python
df1 = spark.createDataFrame( [(1, "Alice", 29), (2, "Bob", 35), (3, "Charlie", 40)], ["id", "name", "age"])
df1.display()
```

```python
 df2 = spark.createDataFrame( [("David", 4, 50), ("Eve", 5, 28), ("Frank", 6, 33)], ["name", "id", "age"] )
 df2.display()
```

```python
result = df1.unionByName(df2).display()
```
