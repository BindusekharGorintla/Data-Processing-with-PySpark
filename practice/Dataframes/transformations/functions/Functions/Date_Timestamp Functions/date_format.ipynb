```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Date and Timestamp Functions").getOrCreate()
```

```python
#date_format(dateExpr: Column, format: String): Column
#Converts a date/timestamp/string to a value of string in the format specified by the date format given by the second argument 
# Note: The output column will be string
```

```python
df = spark.read.table("samples.wanderbricks.employees")
display(df)
```

```python
from pyspark.sql.functions import date_format
df1 = df.withColumn("new_col", date_format(df["joined_at"],"MM-dd-yyyy")).display()
```

```python
df.createOrReplaceTempView("demo")
spark.sql("SELECT date_format(joined_at,'MM-dd-yyyy') as new_col,* FROM demo").display()
```
