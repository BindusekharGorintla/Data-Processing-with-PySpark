```python
# A Left Semi Join in PySpark returns only the rows from the left DataFrame (the first DataFrame mentioned in the join operation) where there is a match with the right DataFrame (the second DataFrame). It does not include any columns from the right DataFrame in the resulting DataFrame. This join type is useful when you only want to filter rows from the left DataFrame based on whether they have a matching key in the right DataFrame
```

```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Joins").getOrCreate()
```

```python
df_p = spark.read.table("healthverity_claims_sample_patient_dataset.hv_claims_sample.provider")
display(df_p)
```

```python
df_e = spark.read.table("healthverity_claims_sample_patient_dataset.hv_claims_sample.enrollment")
display(df_e)
```

```python
from pyspark.sql.functions import regexp_replace
df= df_e.join(df_p, df_e.patient_id == df_p.patient_id, "left_semi")
df.display()
```

```python
df_e.createOrReplaceTempView("df1")
df_p.createOrReplaceTempView("df2")
spark.sql("SELECT * FROM df1 left SEMI JOIN df2 ON df1.patient_id = df2.patient_id").display()
```
