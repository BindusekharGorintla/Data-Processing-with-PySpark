```python
#In PySpark, the collect() action is used to retrieve the entire dataset from a distributed Spark DataFrame or RDD back to the driver program as a local Python object (usually a list)

#It gathers all partitions of the distributed dataset.

#Returns the full dataset to the driver node.

#Typically used for small datasets or when you need to inspect results locally
```

```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName("Actions").getOrCreate()
from pyspark.sql.functions import *
```

```python
data = [("Alice", 25), ("Bob", 30), ("Charlie", 35)]
df = spark.createDataFrame(data, ["Name", "Age"])
display(df)
```

```python
result = df.collect()
print(result)
print(result[0])
print(result[0][1])
```

```python
result = df.collect()
for row in result:
    print(row)
```
