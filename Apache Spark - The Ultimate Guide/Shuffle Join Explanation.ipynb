```python
from pyspark.sql import SparkSession
spark= SparkSession.builder.appName("shuffle join").getOrCreate()
spark.conf.set("spark.sql.join.preferSortMergeJoin", "true")
```

```python
df_e = spark.read.table("healthverity_claims_sample_patient_dataset.hv_claims_sample.enrollment")
df_h = spark.read.table("healthverity_claims_sample_patient_dataset.hv_claims_sample.pharmacy_claim")
```

```python
df = df_e.join((df_h),df_e["patient_id"]==df_h["patient_id"], "inner")
df.display()
```
