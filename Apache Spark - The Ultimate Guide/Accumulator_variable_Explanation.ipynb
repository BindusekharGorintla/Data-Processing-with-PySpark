```python
# Accumulator collect information from workers back to driver typically for counting, sum or tracking metrics
```

```python
from pyspark.sql import SparkSession

# Initialize Spark session
spark = SparkSession.builder.appName("AccumulatorExample").getOrCreate()

```

```python
# Create an accumulator (initial value = 0)
error_count = spark.sparkContext.accumulator(0)


```

```python
# Sample RDD with some values
data = ["1", "2", "three", "4", "five"]
rdd = spark.sparkContext.parallelize(data)

```

```python
# Function that tries to convert to int, increments accumulator if fails
def safe_int(x):
    try:
        return int(x)
    except ValueError:
        error_count.add(1)  # increment accumulator
        return None

# Apply transformation
result = rdd.map(safe_int).collect()

print("Result:", result)
print("Number of errors:", error_count.value)
```
