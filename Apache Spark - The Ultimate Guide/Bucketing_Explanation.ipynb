```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("BucketingExample").getOrCreate()


```

```python
# Sample DataFrame
data = [(1, "Alice"), (2, "Bob"), (3, "Charlie"), (4, "David")]
df = spark.createDataFrame(data, ["id", "name"])

# Write bucketed table
df.write.bucketBy(4, "id").sortBy("id").saveAsTable("workspace.practice.cus_bucket")
```

```python
bucketed_df = spark.table("workspace.practice.cus_bucket") bucketed_df.show()
```
