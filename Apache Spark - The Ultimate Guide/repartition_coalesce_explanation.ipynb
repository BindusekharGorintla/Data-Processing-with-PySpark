```python
from pyspark.sql import SparkSession

# Initialize SparkSession
spark = SparkSession.builder \
    .appName("RepartitionVsCoalesceExample") \
    .getOrCreate()



```

```python
# Let's assume you already have two large DataFrames:
# df_100gb and df_50gb
# Example: Repartitioning a 100 GB DataFrame
# Repartition increases or reshuffles partitions across the cluster
df_100gb_repart = df_100gb.repartition(200)  # e.g., 200 partitions
print("100 GB DataFrame repartitioned to 200 partitions")

```

```python
# Example: Coalescing a 50 GB DataFrame
# Coalesce reduces partitions without full shuffle (more efficient)
df_50gb_coalesce = df_50gb.coalesce(50)  # e.g., reduce to 50 partitions
print("50 GB DataFrame coalesced to 50 partitions")

# Perform some action to trigger execution
df_100gb_repart.count()
df_50gb_coalesce.count()
```
